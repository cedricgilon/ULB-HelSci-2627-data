## 2 · Local or API

> **The idea:** the same kind of model runs on your laptop or at a provider. The response time changes, and so does the place where your data goes

**Choose** where the model runs: `"local"` (Ollama, llama3.2:3b) or `"api"` (Groq, gpt-oss-20b). Run the two cells, then ask again with the other one.
The local model can take 30 to 60 s on a laptop with no GPU.

```python
prompt = "What is the most common side effect of metformin? Answer in two words."
where = "local"
```

```python
answer, seconds = ask(prompt, where, temperature=0)
Markdown(f"**{seconds:.1f} s** · {answer}")
```

- Local: slower with no GPU, the text stays on your laptop
- API: fast, the text goes to the provider

## 3 · Temperature

> **The idea:** the temperature sets how random the draw is. Low: nearly always the most probable token. High: more variety (slide "Temperature")

The same question, 5 times, on the API. How many **different** answers?

**Predict first**, before you run the cells below: a number from 1 to 5.

```python
guess_low = 1   # different answers at temperature 0
guess_high = 1  # different answers at temperature 1.5
```

**Change** the temperature (0 to 1.5), then run the two cells. Target: check your two predictions, then find a temperature where the 5 answers are all different.

```python
temperature = 0
```

```python
answers = [ask(prompt, "api", temperature)[0] for _ in range(5)]
different = len({normalise(answer) for answer in answers})
display(Markdown(f"**{different} different answer(s)** at temperature {temperature}. "
                 f"Your predictions: {guess_low} at 0, {guess_high} at 1.5"))
pd.DataFrame({"answer": answers})
```

**Key points**

- The temperature changes the variety of the answers, not their truth
- At temperature 0, an API can still give a different answer: your request runs in a batch with other requests
- Low temperature for extraction and code, higher for ideas and drafts
