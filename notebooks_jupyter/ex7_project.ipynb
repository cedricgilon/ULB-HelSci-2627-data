## 1 · The prompt and 5 examples

````python
prompt = """
Say whether a sentence reports an adverse event: yes or no.
Answer with one word.
"""
````

````python
examples = """
After the second dose, the patient developed a rash on both arms.
The patient took the tablets with breakfast, as advised.
She reported dizziness two hours after the injection.
The study enrolled 120 adults with type 2 diabetes.
Three participants stopped treatment because of nausea.
"""
````

## 2 · Run on the 5 examples

> **The idea:** the instructions on one side, the data between tags on the other (slide "Data and messages")

The notebook sends one message per example: your prompt first, then the example between `<input>` tags. **Read the code cells** (no change needed) and run them: first the message for your first example, then the answers.

````python
def make_message(prompt, example):
    """The prompt, then the example between <input> tags."""
    return f"{prompt}\n\n<input>\n{example}\n</input>"
````

````python
example_list = [line.strip() for line in examples.splitlines() if line.strip()]
if example_list:
    preview = Markdown(f"```\n{make_message(prompt.strip(), example_list[0])}\n```")
else:
    preview = Markdown("Write at least one example, one per line.")
preview
````

`where`: `"api"` (Groq) or `"local"` (Ollama).

````python
where = "api"
````

````python
outputs = []
for example in example_list:
    response = client(where).chat.completions.create(
        model=PROVIDERS[where]["model"],
        messages=[{"role": "user", "content": make_message(prompt.strip(), example)}],
        temperature=0,
    )
    outputs.append(response.choices[0].message.content)
````

````python
pd.DataFrame({"example": example_list, "answer": outputs})
````
